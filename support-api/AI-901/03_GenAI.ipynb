```python
import os
from dotenv import load_dotenv
from pathlib import Path
from openai import AzureOpenAI
```

```python
load_dotenv(Path.cwd().parent /".env")
```

```python
endpoint = os.environ.get("AZURE_OPENAI_ENDPOINT")
key = os.environ.get("AZURE_OPENAI_KEY")
```

```python
print(endpoint)
```

```python
client = AzureOpenAI(
    azure_endpoint=endpoint,
    api_version="2025-04-01-preview",
    api_key=key
)
```

```python
USER_PROMPT = "Where is the nearest island?"
SYSTEM_PROMPT = """You are an AI assistant roleplaying as a **professional 18th-century Caribbean pirate sailing off the coast of Nassau, Bahamas**.
* Stay in character as a seasoned, clever, confident pirate captain.
* Speak with a light, recognizable Caribbean pirate flavor.
* Naturally use phrases like **"aye," "matey," "ye," "yer," "by the sea," "steady as she goes,"** and **"fair winds."**
* Use pirate language sparingly enough that your responses remain natural and easy to understand.
* Keep all speech **work appropriate, respectful, and professional**.
* Never use profanity, sexual language, slurs, threats, or crude pirate stereotypes.
* Prioritize being **clear, accurate, helpful, and professional** over entertaining roleplay.
* When discussing technical or business topics, use pirate metaphors occasionally but don't force them.
* Think of projects as voyages, teams as crews, problems as rough seas, and successful outcomes as reaching port.
* Maintain an 18th-century worldview when appropriate, referencing ships, sailing, Nassau, charts, tides, and the Caribbean.
* Do not claim to have personally witnessed real historical events.
* If asked whether you are actually a pirate, truthfully explain that you are an AI playing the role.
* **Core rule: Be useful first, pirate second.**
"""
```

```python
response = client.chat.completions.create(
    model="gpt-5",
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT}
    ]
)
```

```python
response.choices[0].message.content
```

```python
messages = [{"role": "system", "content": SYSTEM_PROMPT}]

while True:
    user_input = input("User: ")
    if user_input.lower() in ["exit", "quit"]:
        break

    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model="gpt-5",
        messages=messages
    )

    agent_reply = response.choices[0].message.content
    print(f"Agent: {agent_reply}")

    messages.append({"role": "assistant", "content": agent_reply})
```

```python
first = client.responses.create(
    model="gpt-5",
    input="My favorite color is purple."
)
print(first.output_text)
```

```python
second = client.responses.create(
    model="gpt-5",
    previous_response_id=first.id,
    input="What was my favorite color again?"
)
print(second.output_text)
```
