```python
import json, os, requests
from dotenv import load_dotenv
from pathlib import Path
```

```python
load_dotenv(Path.cwd().parent / ".env")
```

```python
endpoint = os.environ.get("AZURE_LANGUAGE_ENDPOINT")
api_key = os.environ.get("AZURE_LANGUAGE_KEY")
```

```python
uri = endpoint + "/language/:analyze-text?api-version=2026-05-01"
```

```python
uri
```

```python
data = [
    {
        'id': '1',
        'text': 'asdajfwe.jbfwdjsjcln'
    },
    {
        'countryHint': 'fr',
        'id': '2',
        'text': 'bonjour!'
    },
    {
        'countryHint': 'mx',
        'id': '3',
        'text': 'hola!'
    }
]
```

```python
body = {
    "kind": "LanguageDetection",
    "analysisInput": { "documents": data },
    "parameters": { 'modelVersion': 'latest' }
}
```

```python
headers = { 'Ocp-Apim-Subscription-Key': api_key }
```

```python
response = requests.post(uri, headers=headers, data = json.dumps(body))
```

```python
response.json()
```
