```python
from azure.ai.textanalytics import TextAnalyticsClient
from azure.core.credentials import AzureKeyCredential

import os
from dotenv import load_dotenv
from pathlib import Path
import json
```

```python
load_dotenv(Path.cwd().parent /".env")
```

```python
endpoint = os.environ.get("AZURE_LANGUAGE_ENDPOINT")
key = os.environ.get("AZURE_LANGUAGE_Key")
```

```python
credential = AzureKeyCredential(key)
client = TextAnalyticsClient(
    endpoint=endpoint,
    credential=credential
)
```

```python
txt_path = Path.cwd().parent / "data" / "tickets.json"
```

```python
txt_path # C:\Users\BrandonVanek\Desktop\AIDev\20260713 Guidehouse\support-platform\support-api\data\tickets.json
```

```python
txt_files = json.load(open(txt_path, "r"))
```

```python
txt_files
```

```python
data_dict = {ticket.get("id", None): ticket.get("title", "") for ticket in txt_files}
```

```python
data_dict
```

```python
detect_language = client.detect_language(documents=[data_dict["TKT-10001"]])
```
